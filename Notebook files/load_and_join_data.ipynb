```python
import pandas as pd
import os
```

```python
DATA_DIR = "../Data"
OUTPUT_DIR = "../Merged_Olist_Data"
os.makedirs(OUTPUT_DIR, exist_ok=True)
```

Loading CSV Files

```python
df_orders    = pd.read_csv(os.path.join(DATA_DIR, "olist_orders_dataset.csv"))
df_reviews   = pd.read_csv(os.path.join(DATA_DIR, "olist_order_reviews_dataset.csv"))
df_customers = pd.read_csv(os.path.join(DATA_DIR, "olist_customers_dataset.csv"))
df_products  = pd.read_csv(os.path.join(DATA_DIR, "olist_products_dataset.csv"))


print(f"Orders:    {df_orders.shape}")
print(f"Reviews:   {df_reviews.shape}")
print(f"Customers: {df_customers.shape}")
print(f"Products:  {df_products.shape}")
```

```python
df_products.head()




df_orders.head()
df_reviews.head()
df_customers.head()
```

```python
df_reviews = (df_reviews
              .sort_values("review_creation_date")
              .drop_duplicates("order_id", keep="last"))

df_merged = df_orders.merge(df_reviews, on="order_id", how="left")
```

```python
df_merged = df_merged.merge(df_customers, on="customer_id", how="left")
```

```python
print(f"Original orders row count : {len(df_orders)}")
print(f"After joins row count     : {len(df_merged)}")

if len(df_merged) == len(df_orders):
    print("Row count matches")
else:
    diff = len(df_merged) - len(df_orders)
    print(f"Row count increased by {diff}")
```

```python
print("Shape:", df_merged.shape)
print("Columns:", df_merged.columns.tolist())
df_merged.head()
```

```python
output_path = os.path.join(OUTPUT_DIR, "master_orders.csv")
df_merged.to_csv(output_path, index=False)
print(f"master_orders.csv saved to {output_path}")
```
